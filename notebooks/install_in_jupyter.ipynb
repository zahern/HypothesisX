# Installing SearchLibrium in Jupyter Notebook

You can install SearchLibrium directly from within a Jupyter notebook cell using `subprocess` and `sys`.

````python
# Install SearchLibrium
import subprocess
import sys

subprocess.check_call([sys.executable, "-m", "pip", "install", "SearchLibrium", "--upgrade"])
````

````python
# Verify installation
from SearchLibrium import Parameters, call_siman
import SearchLibrium

print(f"✓ SearchLibrium {SearchLibrium.__version__} installed successfully!")
````

## Now you can use SearchLibrium in your notebook:

```python
import numpy as np
import pandas as pd

# Load your data
df = pd.read_csv("your_data.csv")
varnames = ["TIME", "COST", "HEADWAY"]
choice_set = ["car", "train", "bus"]

# Create search parameters
params = Parameters(
    criterions=[("bic", -1)],
    df=df,
    varnames=varnames,
    asvarnames=varnames,
    choice_set=choice_set,
    choices=df["CHOICE"].values,
    alt_var=df["alt"].values,
    choice_id=df["choice_id"].values,
    base_alt="car",
    models=["multinomial", "mixed_logit"],
    p_val=0.05,
)

# Run model search
best = call_siman(params, init_sol=None, id_num=1)
```
